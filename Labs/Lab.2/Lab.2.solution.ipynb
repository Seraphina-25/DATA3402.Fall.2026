*Exercise 1:* Write a function that creates an **n×n matrix** (a list of lists) representing the state of a tic‑tac‑toe game.

Use the integers:

- `0` = empty
- `1` = `"X"`
- `2` = `"O"`


````python
# Write your solution here
def create_board(n):
    board = []

    for i in range(n):
        row = []

        for j in range(n):
            row.append(" ")

        board.append(row)

    return board
````

*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def check_game(board):
    n = len(board)

    for row in board:
        if all(cell == "X" for cell in row):
            return 1
        if all(cell == "O" for cell in row):
            return 2

    for j in range(n):
        if all(board[i][j] == "X" for i in range(n)):
            return 1
        if all(board[i][j] == "O" for i in range(n)):
            return 2

    if all(board[i][i] == "X" for i in range(n)):
        return 1
    if all(board[i][i] == "O" for i in range(n)):
        return 2

    if all(board[i][n - 1 - i] == "X" for i in range(n)):
        return 1
    if all(board[i][n - 1 - i] == "O" for i in range(n)):
        return 2

    for row in board:
        if " " in row:
            return -1

    return 0
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Write your solution 
def make_move(board, player, row, col):

    if board[row][col] != " ":
        return False

    if player == 1:
        board[row][col] = "X"
    elif player == 2:
        board[row][col] = "O"
    else:
        return False

    return True
````

````python
# Test your solution 
board = [
    [" ", " ", " "],
    [" ", " ", " "],
    [" ", " ", " "]
]

result = make_move(board, 1, 1, 1)

print(result)
print(board)
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
# Write your solution here
def move_location(board, player, location):
    row = ord(location[0].upper()) - ord("A")
    col = int(location[1:]) - 1

    return make_move(board, player, row, col)
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
# Write yourrr solution here
def play_game(n):
    board = create_board(n)
    player = 1

    while True:
        draw_board(board)

        player_move(board, player)

        status = check_game(board)

        if status == 1:
            draw_board(board)
            print("Player 1 (X) wins!")
            break

        elif status == 2:
            draw_board(board)
            print("Player 2 (O) wins!")
            break

        elif status == 0:
            draw_board(board)
            print("The game is a draw!")
            break

        # Switch players
        if player == 1:
            player = 2
        else:
            player = 1
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
def computer_move(board, player):
    n = len(board)
    mark = "O" if player == 2 else "X"

    # Try to win
    for row in range(n):
        for col in range(n):
            if board[row][col] == " ":
                board[row][col] = mark

                if check_game(board) == player:
                    return True

                board[row][col] = " "

    # Try to block the other player
    other_player = 1 if player == 2 else 2
    other_mark = "X" if other_player == 1 else "O"

    for row in range(n):
        for col in range(n):
            if board[row][col] == " ":
                board[row][col] = other_mark

                if check_game(board) == other_player:
                    board[row][col] = mark
                    return True

                board[row][col] = " "

    # Otherwise choose an empty space
    for row in range(n):
        for col in range(n):
            if board[row][col] == " ":
                board[row][col] = mark
                return True

    return False


def play_computer_game(n):
    board = create_board(n)
    player = 1

    while True:
        draw_board(board)

        if player == 1:
            print("Your turn (X)")
            player_move(board, player)
        else:
            print("Computer's turn (O)")
            computer_move(board, player)

        status = check_game(board)

        if status == 1:
            draw_board(board)
            print("Player 1 (X) wins!")
            break
        elif status == 2:
            draw_board(board)
            print("Computer (O) wins!")
            break
        elif status == 0:
            draw_board(board)
            print("The game is a draw!")
            break

        if player == 1:
            player = 2
        else:
            player = 1
````

````python
# Test your solution here
play_computer_game(3)
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here
def minimax(board, depth, max_depth, computer_turn):
    result = check_game(board)

    if result == 2:
        return 1
    if result == 1:
        return -1
    if result == 0 or depth == max_depth:
        return 0

    scores = []

    for r in range(len(board)):
        for c in range(len(board)):
            if board[r][c] == " ":
                board[r][c] = "O" if computer_turn else "X"

                score = minimax(board, depth + 1, max_depth,
                                not computer_turn)

                board[r][c] = " "
                scores.append(score)

    return max(scores) if computer_turn else min(scores)


def computer_move(board, max_depth):
    best_score = -100
    best_move = None

    for r in range(len(board)):
        for c in range(len(board)):
            if board[r][c] == " ":
                board[r][c] = "O"
                score = minimax(board, 1, max_depth, False)
                board[r][c] = " "

                if score > best_score:
                    best_score = score
                    best_move = (r, c)

    if best_move:
        r, c = best_move
        board[r][c] = "O"


def play_computer_game(n, max_depth):
    board = create_board(n)

    while True:
        draw_board(board)

        player_move(board, 1)

        if check_game(board) != -1:
            break

        computer_move(board, max_depth)

        if check_game(board) != -1:
            break

    draw_board(board)

    result = check_game(board)

    if result == 1:
        print("Player wins!")
    elif result == 2:
        print("Computer wins!")
    else:
        print("Draw!")
        
````

````python
# Test your solution here
play_computer_game(3, 5)
````
